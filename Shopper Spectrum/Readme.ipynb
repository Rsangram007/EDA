# 🛒 Shopper Spectrum: Customer Segmentation and Product Recommendations in E-Commerce

## 📌 Project Overview (What)

This project analyzes transactional data from an online retail store to achieve two main objectives:

1. **Customer Segmentation** using RFM (Recency, Frequency, Monetary) analysis combined with K-Means clustering.
2. **Product Recommendation System** using Item-based Collaborative Filtering.

The outcome is an interactive **Streamlit web application** that allows users to:
- Get personalized product recommendations by entering a product name.
- Predict a customer's segment by inputting their RFM values.

The project demonstrates end-to-end data science workflow: data cleaning, exploratory analysis, feature engineering, unsupervised machine learning, recommendation systems, model persistence, and deployment of an interactive dashboard.

## 🎯 Why This Project? (Problem Statement & Importance)

The e-commerce industry generates massive transaction data every day. Without proper analysis, businesses miss opportunities to:

This project solves these by turning raw transaction data into **actionable customer segments** and **personalized product suggestions**.

## 🛠 How It Works (Technical Approach & Methodology)

### Dataset
- **Source**: UCI Online Retail Dataset (transactions from 2022–2023, primarily UK-based gift retailer).
- **Key Columns**: InvoiceNo, StockCode, Description, Quantity, InvoiceDate, UnitPrice, CustomerID, Country.

### Step-by-Step Process

1. **Data Preprocessing**
   - Removed cancelled orders (InvoiceNo starting with 'C').
   - Dropped rows with missing CustomerID.
   - Filtered out negative/zero Quantity and UnitPrice.
   - Created `TotalPrice = Quantity × UnitPrice`.

2. **Exploratory Data Analysis (EDA)**
   - Transaction volume by country (UK dominates).
   - Top-selling products (mostly decorative/home items).
   - Revenue trends over time.
   - Distribution of customer spending.

3. **Customer Segmentation (RFM + K-Means)**
   - **Recency**: Days since last purchase.
   - **Frequency**: Number of unique invoices per customer.
   - **Monetary**: Total spend per customer.
   - Standardized RFM values.
   - Applied K-Means clustering (optimal k=4 via Elbow & Silhouette methods).
   - Labeled segments:
     - **High-Value**: Recent, frequent, high-spending (VIP customers).
     - **Regular**: Steady purchasers.
     - **Occasional**: Infrequent, low spend.
     - **At-Risk**: Long time since last purchase, low activity.

4. **Product Recommendation System**
   - Created a Product × Customer pivot table (Quantity as implicit rating).
   - Computed **cosine similarity** between products.
   - For any input product, returns top-5 most similar products (item-based collaborative filtering).

5. **Model Persistence & Deployment**
   - Saved trained models (KMeans, StandardScaler, similarity matrix, cluster labels).
   - Built interactive **Streamlit app** for real-time predictions.

### Tech Stack
- **Language**: Python
- **Libraries**: Pandas, NumPy, Matplotlib, Seaborn, Scikit-learn, Streamlit, Pickle
- **Environment**: Databricks Community Edition (notebook execution)
- **Deployment**: Local Streamlit app

## 📊 Key Business Insights from the Analysis

Based on typical results from this dataset:

1. **Geographic Concentration**
   - ~90% of transactions from the United Kingdom → Focus marketing budget on UK, explore international growth carefully.

2. **Product Trends**
   - Top products are low-cost decorative/gift items (e.g., "WHITE HANGING HEART T-LIGHT HOLDER", lunch boxes, cake cases).
   - High similarity among seasonal and themed items → Great opportunity for bundling (e.g., Christmas decorations together).

3. **Customer Behavior Patterns**
   - Small group of **High-Value customers** contribute disproportionately to revenue (Pareto principle).
   - Many **At-Risk customers** → Retention campaigns could significantly boost revenue.
   - **Occasional buyers** dominate in count but not in value → Convert them to Regular via recommendations.

4. **Revenue Drivers**
   - Peak sales around Christmas season → Prepare inventory and marketing well in advance.
   - Frequent small orders → Encourage bundling to increase average order value.

5. **Recommendation Power**
   - Products bought together are often thematically similar (e.g., party items, kitchenware sets).
   - Implementing these recommendations can directly increase cross-sell revenue.
